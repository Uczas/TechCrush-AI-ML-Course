```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
```

```python
print("--- Phase 1: Object Detection (Bounding Boxes) ---")
det_model = YOLO('yolov8n.pt')

print("Scanning image for objects...")
det_results = det_model('ozoro_market.jpg')

det_image = det_results[0].plot()
```

```python
print("\n--- Phase 2: Image Segmentation (Pixel Masks) ---")
seg_model = YOLO('yolov8n-seg.pt')

print("Tracing exact objects boundaries....")
seg_results = seg_model('ozoro_market.jpg')

seg_image = seg_results[0].plot()
```

```python
print("\n--- Phase 3: Visualizing the AI's Vision ---")
det_image_rgb = cv2.cvtColor(det_image, cv2.COLOR_BGR2RGB)
seg_image_rgb = cv2.cvtColor(seg_image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 10))

plt.subplot(1, 2, 1)
plt.title("Object Detection (Bounding Boxes)", fontsize=10, fontweight='bold')
plt.imshow(det_image_rgb)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Image Segmentation (Pixel Masks)", fontsize=10, fontweight='bold')
plt.imshow(seg_image_rgb)
plt.axis('off')


plt.tight_layout()
plt.show()
```
